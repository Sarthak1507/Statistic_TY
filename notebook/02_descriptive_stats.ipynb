```python
"""
Script 2: Descriptive Statistics
Calculate mean, median, standard deviation, variance, skewness, kurtosis, etc.
"""

import pandas as pd
import numpy as np
from scipy import stats
from pathlib import Path

def calculate_descriptive_stats(df, column):
    """Calculate comprehensive descriptive statistics"""
    data = df[column].dropna()
    
    stats_dict = {
        'Column': column,
        'Count': len(data),
        'Mean': data.mean(),
        'Median': data.median(),
        'Mode': data.mode().values[0] if len(data.mode()) > 0 else np.nan,
        'Std Dev': data.std(),
        'Variance': data.var(),
        'Min': data.min(),
        'Max': data.max(),
        'Range': data.max() - data.min(),
        'Q1 (25%)': data.quantile(0.25),
        'Q3 (75%)': data.quantile(0.75),
        'IQR': data.quantile(0.75) - data.quantile(0.25),
        'Skewness': data.skew(),
        'Kurtosis': data.kurtosis(),
        'CV (%)': (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan,
    }
    
    return stats_dict

def generate_statistics_report(df):
    """Generate complete statistics report"""
    print("\n" + "="*80)
    print("DESCRIPTIVE STATISTICS REPORT")
    print("="*80)
    
    numeric_columns = df.select_dtypes(include=[np.number]).columns
    
    all_stats = []
    
    for col in numeric_columns:
        if col != 'No':  # Skip the index column
            stats_dict = calculate_descriptive_stats(df, col)
            all_stats.append(stats_dict)
            
            print(f"\n{'─'*80}")
            print(f"Column: {col}")
            print(f"{'─'*80}")
            for key, value in stats_dict.items():
                if key != 'Column':
                    if isinstance(value, float):
                        print(f"  {key:.<25} {value:>15.4f}")
                    else:
                        print(f"  {key:.<25} {value:>15}")
    
    # Create DataFrame
    stats_df = pd.DataFrame(all_stats)
    
    # Save to CSV
    Path('outputs').mkdir(exist_ok=True)
    stats_df.to_csv('outputs/descriptive_statistics.csv', index=False)
    print(f"\n✓ Statistics saved to outputs/descriptive_statistics.csv")
    
    return stats_df

def calculate_percentiles(df, column, percentiles=[5, 10, 25, 50, 75, 90, 95]):
    """Calculate percentiles"""
    data = df[column].dropna()
    
    print(f"\n{'─'*50}")
    print(f"PERCENTILES FOR {column}")
    print(f"{'─'*50}")
    
    for p in percentiles:
        value = np.percentile(data, p)
        print(f"  {p}th Percentile: {value:.4f}")
    
    return data

if __name__ == "__main__":
    # Load cleaned data
    file_path = 'outputs/cleaned_data.csv'
    
    try:
        df = pd.read_csv(file_path)
        df['Date'] = pd.to_datetime(df['Date'])
        
        # Generate statistics report
        stats_df = generate_statistics_report(df)
        
        # Calculate percentiles for Close Price
        calculate_percentiles(df, 'Close Price')
        
        print("\n✓ Descriptive statistics analysis completed!")
        
    except FileNotFoundError:
        print(f"Please run 01_data_loading.py first to generate {file_path}")
```
