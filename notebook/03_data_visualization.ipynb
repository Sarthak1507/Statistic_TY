```python
"""
Script 3: Data Visualization
Create plots and visualizations for the data
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)

def create_output_dirs():
    """Create output directories"""
    Path('outputs').mkdir(exist_ok=True)
    Path('outputs/plots').mkdir(exist_ok=True)

def plot_close_price_trend(df):
    """Plot closing price trend over time"""
    fig, ax = plt.subplots(figsize=(14, 6))
    
    ax.plot(df['Date'], df['Close Price'], linewidth=2, color='#2E86AB', label='Close Price')
    ax.fill_between(df['Date'], df['Close Price'], alpha=0.3, color='#2E86AB')
    
    ax.set_xlabel('Date', fontsize=12, fontweight='bold')
    ax.set_ylabel('Price', fontsize=12, fontweight='bold')
    ax.set_title('Stock Close Price Trend Over Time', fontsize=14, fontweight='bold')
    ax.legend(fontsize=10)
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    plt.savefig('outputs/plots/01_price_trend.png', dpi=300, bbox_inches='tight')
    print("✓ Saved: 01_price_trend.png")
    plt.close()

def plot_moving_averages(df):
    """Plot moving averages"""
    fig, ax = plt.subplots(figsize=(14, 6))
    
    ax.plot(df['Date'], df['Close Price'], linewidth=1.5, label='Close Price', alpha=0.7, color='#2E86AB')
    ax.plot(df['Date'], df['3 Day MV.'], linewidth=2, label='3-Day Moving Average', color='#A23B72')
    ax.plot(df['Date'], df['5 Day MV.'], linewidth=2, label='5-Day Moving Average', color='#F18F01')
    
    ax.set_xlabel('Date', fontsize=12, fontweight='bold')
    ax.set_ylabel('Price', fontsize=12, fontweight='bold')
    ax.set_title('Moving Averages Comparison', fontsize=14, fontweight='bold')
    ax.legend(fontsize=10)
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    plt.savefig('outputs/plots/02_moving_averages.png', dpi=300, bbox_inches='tight')
    print("✓ Saved: 02_moving_averages.png")
    plt.close()

def plot_distribution(df, column):
    """Plot distribution of a column"""
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    data = df[column].dropna()
    
    # Histogram
    axes[0].hist(data, bins=30, color='#2E86AB', edgecolor='black', alpha=0.7)
    axes[0].set_xlabel(column, fontsize=12, fontweight='bold')
    axes[0].set_ylabel('Frequency', fontsize=12, fontweight='bold')
    axes[0].set_title(f'Distribution of {column}', fontsize=12, fontweight='bold')
    axes[0].axvline(data.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {data.mean():.2f}')
    axes[0].legend()
    
    # Box plot
    axes[1].boxplot(data, vert=True)
    axes[1].set_ylabel(column, fontsize=12, fontweight='bold')
    axes[1].set_title(f'Box Plot of {column}', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(f'outputs/plots/03_distribution_{column.replace(" ", "_")}.png', dpi=300, bbox_inches='tight')
    print(f"✓ Saved: 03_distribution_{column.replace(' ', '_')}.png")
    plt.close()

def plot_qq_plot(df, column):
    """Plot Q-Q plot for normality assessment"""
    from scipy import stats as scipy_stats
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    data = df[column].dropna()
    scipy_stats.probplot(data, dist="norm", plot=ax)
    ax.set_title(f'Q-Q Plot for {column}', fontsize=14, fontweight='bold')
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(f'outputs/plots/04_qq_plot_{column.replace(" ", "_")}.png', dpi=300, bbox_inches='tight')
    print(f"✓ Saved: 04_qq_plot_{column.replace(' ', '_')}.png")
    plt.close()

def plot_price_comparison(df):
    """Create comprehensive price comparison plot"""
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # Plot 1: Price trend
    axes[0, 0].plot(df['Date'], df['Close Price'], color='#2E86AB', linewidth=2)
    axes[0, 0].set_title('Close Price Trend', fontweight='bold')
    axes[0, 0].set_xlabel('Date')
    axes[0, 0].set_ylabel('Price')
    axes[0, 0].grid(True, alpha=0.3)
    
    # Plot 2: Daily returns
    df['Daily_Return'] = df['Close Price'].pct_change() * 100
    axes[0, 1].bar(df['Date'], df['Daily_Return'], color='#A23B72', alpha=0.7)
    axes[0, 1].set_title('Daily Returns (%)', fontweight='bold')
    axes[0, 1].set_xlabel('Date')
    axes[0, 1].set_ylabel('Return (%)')
    axes[0, 1].grid(True, alpha=0.3)
    
    # Plot 3: Price volatility (rolling std)
    df['Volatility'] = df['Close Price'].rolling(window=20).std()
    axes[1, 0].plot(df['Date'], df['Volatility'], color='#F18F01', linewidth=2)
    axes[1, 0].set_title('20-Day Rolling Volatility', fontweight='bold')
    axes[1, 0].set_xlabel('Date')
    axes[1, 0].set_ylabel('Std Dev')
    axes[1, 0].grid(True, alpha=0.3)
    
    # Plot 4: Distribution of prices
    axes[1, 1].hist(df['Close Price'].dropna(), bins=30, color='#2E86AB', edgecolor='black', alpha=0.7)
    axes[1, 1].set_title('Price Distribution', fontweight='bold')
    axes[1, 1].set_xlabel('Price')
    axes[1, 1].set_ylabel('Frequency')
    axes[1, 1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('outputs/plots/05_comprehensive_analysis.png', dpi=300, bbox_inches='tight')
    print("✓ Saved: 05_comprehensive_analysis.png")
    plt.close()

if __name__ == "__main__":
    create_output_dirs()
    
    file_path = 'outputs/cleaned_data.csv'
    
    try:
        df = pd.read_csv(file_path)
        df['Date'] = pd.to_datetime(df['Date'])
        
        print("\n" + "="*70)
        print("CREATING VISUALIZATIONS")
        print("="*70)
        
        plot_close_price_trend(df)
        plot_moving_averages(df)
        plot_distribution(df, 'Close Price')
        plot_qq_plot(df, 'Close Price')
        plot_price_comparison(df)
        
        print("\n✓ All visualizations completed!")
        print("  Check outputs/plots/ directory for all plots")
        
    except FileNotFoundError:
        print(f"Please run 01_data_loading.py first to generate {file_path}")
```
